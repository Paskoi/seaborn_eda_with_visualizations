### Duplicates

```python
df_titanic.duplicated().sum()
```

```python
df_titanic[df_titanic.duplicated(keep=False)]
```

```python
dup_count = df_titanic.duplicated().sum()
total = len(df_titanic)

print(f"Duplicates: {dup_count}/{total} ({dup_count / total:.2%}) of the dataset")
```

```python
dup_mask = df_titanic.duplicated(keep=False)

print(df_titanic[dup_mask]["pclass"].value_counts(normalize=True).round(2))
print(df_titanic["pclass"].value_counts(normalize=True).round(2))
print(df_titanic[dup_mask]["alone"].mean().round(2), df_titanic["alone"].mean().round(2))
```

The dataset has no passenger ID, name or ticket number, so identical rows cannot be confirmed as the same person. Rows in duplicate groups are mostly third-class (78% vs 55% in the whole data) and travelling alone (83% vs 60%), which fits a common profile rather than data entry errors. All rows will be kept.

### Missing values and categories

```python
df_titanic.isnull().sum()
```

```python
df_titanic.isna().sum()
```

```python
print(df_titanic[df_titanic["embarked"].isna()])
```

```python
age_missing = df_titanic["age"].isna().sum()
deck_missing = df_titanic["deck"].isna().sum()
total = len(df_titanic)

print(f"Age missing: {age_missing}/{total} ({age_missing / total:.2%}) of the dataset")
print(f"Deck missing: {deck_missing}/{total} ({deck_missing / total:.2%}) of the dataset")
```

```python
missing_by_class = (
    df_titanic.assign(age_missing=df_titanic["age"].isna())
    .groupby("class")["age_missing"]
    .agg(missing_count="sum", total="count", missing_pct="mean")
)

print(missing_by_class)
```

```python
for col in df_titanic.select_dtypes(include="category"):
    print(f"{col}: {df_titanic[col].unique().tolist()}")

for col in df_titanic.select_dtypes(include="str"):
    print(f"{col}: {df_titanic[col].unique().tolist()}")

print(df_titanic[["sex", "embarked", "class", "who", "adult_male", "embark_town", "alive", "alone"]].nunique())
```

- `embarked` / `embark_town`: 2 rows are missing. Both are first-class women who survived. The rows are kept in the data and excluded only from the port analysis, so the other questions still use all 891 passengers.
- `age`: 177 of 891 rows are missing (19.9%). 27.7% of third-class passengers have no age, compared with 13.9% in first class and 6.0% in second class. The values are neither dropped nor filled. Analyses that use age are run on rows with a known age only, so third-class passengers will be underrepresented there.
- `deck`: 688 of 891 rows are missing (77.2%). Too sparse to be useful, so the column is not used in the analysis.
